```python
# import libraries
import pandas as pd
import csv
from titlecase import titlecase
import matplotlib.pyplot as plt
from collections import Counter
```

```python
# load data
data = pd.read_excel(r"../Dataset/superstore.xls")
```

```python
# data overview
print("=" * 40)
print("DATASET OVERVIEW")

# rows and columns 
print("=" * 40)
rows, cols = data.shape
print(f"Jumlah Baris  : {rows:,}")
print(f"Jumlah Kolom  : {cols}")

# column names
print("=" * 40)
print("Nama Kolom : ")
for index, kolom in enumerate (data.columns, start = 1) :
    print(f"{index}. {kolom}")

# data types
print("=" * 40)
print("Nama dan Tipe Data Kolom : ")
df_dtypes = pd.DataFrame({
    "Nama kolom" : data.dtypes.index,
    "Tipe Data" : data.dtypes.values
})
display(df_dtypes.style.hide(axis = "index"))

# preview dataset
print("=" * 40)
print("Preview Data : ")
display(data.head(3))
```

```python
# missing values 
print("=" * 40)
print("MISSING VALUES")

# check null values per columns
print("=" * 40)
display(pd.DataFrame(
    {"Nama Kolom" : data.isna().sum().index,
    "Jumlah Null" : data.isna().sum().values}
).style.hide(axis = "index"))
```

```python
# duplicate values 
print("=" * 40)
print("DUPLICATE RECORDS")

# check duplicate values
print("=" * 40)
print(f"Data duplikat : {data.duplicated().sum()}")
```

```python
print("=" * 40)
print("TEXT STANDARDIZATION")

# string columns list trim
string_columns_trim = ["Order ID", "Ship Mode", "Customer ID", "Customer Name", "Segment", "Country/Region", "City", "State/Province", 
                       "Region", "Product ID", "Category", "Sub-Category", "Product Name"]

# trim leading and trailing whitespace from a string
print("=" * 40)
for i in string_columns_trim :
    data[i]=data[i].str.strip()

# string columns list
string_columns = ["Ship Mode", "Customer Name", "Segment", "Country/Region", "City", "State/Province", "Region", 
                  "Category", "Sub-Category", "Product Name"]

# standardizing letter case
for i in string_columns :
    data[i] = data[i].apply(titlecase)
```

```python
print("=" * 40)
print("DATA TYPE")

print("=" * 40)
print("Preview Data Before Change Dtypes ")

# order date and ship date column alr datetime types

# change the data type of the ship mode column to category
# check ship mode 
print(f"1. Ship Mode : {data["Ship Mode"].unique()}")
# change to category
data["Ship Mode"] = data["Ship Mode"].astype('category')

# change the data type of the segment column to category
# check segment 
print(f"2. Segment : {data["Segment"].unique()}")
# change to category
data["Segment"] = data["Segment"].astype('category')

# change the data type of the country/region column to category
# check country/region 
print(f"3. Country/Region : {data["Country/Region"].unique()}")
# change to category
data["Country/Region"] = data["Country/Region"].astype('category')

# change the data type of the city column to category
# check city 
print(f"4. City : {data["City"].unique()}")
# change to category
data["City"] = data["City"].astype('category')

# change the data type of the state/province column to category
# check state/province
print(f"5. State/Province : {data["State/Province"].unique()}")
# change to category
data["State/Province"] = data["State/Province"].astype('category')

# change the data type of the region column to category
# check region 
print(f"6. Region : {data["State/Province"].unique()}")
# change to category
data["Region"] = data["Region"].astype('category')

# change the data type of the category column to category
# check category 
print(f"7. Category : {data["Category"].unique()}")
# change to category
data["Category"] = data["Category"].astype('category')

# change the data type of the sub-category column to category
# check sub-category 
print(f"8. Sub-Category : {data["Sub-Category"].unique()}")
# change to category
data["Sub-Category"] = data["Sub-Category"].astype('category')

# change the data type of the Product Name column to category
# check sub-category 
print(f"9. Product Name : {data["Product Name"].unique()}")
# change to category
data["Product Name"] = data["Product Name"].astype('category')

print("=" * 40)
print("Preview Data After Change Dtypes ")
display(pd.DataFrame({
    "Nama kolom" : data.dtypes.index,
    "Tipe Data" : data.dtypes.values
}).style.hide(axis = "index"))
```

```python
print("=" * 40)
print("DATA VALIDATION")

# checking for anomalous values 
print("=" * 40)
print(f"Jumlah data sales < 0 : {len(data[data["Sales"] < 0])}")
print(f"Jumlah data diskon < 0 : {len(data[data["Discount"] < 0])}")
```

```python
# replace " in middle string to '
# Hapus tanda " di awal dan akhir string
data["Product Name"] = data["Product Name"].str.strip('"')

# Ganti tanda " yang berada di tengah menjadi '
data["Product Name"] = data["Product Name"].str.replace(
    r'(?<=.)"(?=.)',
    "'",
    regex=True
)
```

```python
# rename country/region data 
data.rename(columns={
    "Country/Region": "Country_Region", 
    "State/Province" : "State_Province"
}, inplace=True)
```

```python
# save to csv
data.to_csv("../Clean Data/superstore_cleaned.csv", index = False, encoding = "utf-8")
print("Data berhasil disimpan di folder Clean Data.")
```
